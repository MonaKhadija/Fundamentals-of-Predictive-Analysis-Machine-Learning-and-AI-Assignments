**Dataset and Problem Statement:**
We will be sing the CIFAR-10 dataset to to build a convolutional neural network (CNN). First, we will Change runtime type, select T4 GPU to optimize our speed.

**1. Environment Setup:**

```python
# Importing libraries
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, random_split
import torchvision.transforms as transforms
import csv
import numpy as np
import os
import time
```

```python
# Checking GPU availability
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

**2. settig up Data Loader:**

```python
class RobustCSVDataset(Dataset):
    def __init__(self, csv_file, transform=None):
        self.transform = transform
        self.labels = []
        self.pixels = []

        print("Reading CSV line by line...")
        with open(csv_file, 'r', encoding='utf-8') as f:
            reader = csv.reader(f)
            for row in reader:
                if not row or len(row) < 2:
                    continue
                try:
                    # First column is label, remaining are pixels
                    label = int(row[0])
                    pix = [float(p) for p in row[1:]]

                    self.labels.append(label)
                    self.pixels.append(pix)
                except ValueError:
                    continue

        # Clipping labels to 0-9 to prevent CUDA out-of-bounds assertions
        self.labels = np.clip(np.array(self.labels, dtype=np.int64), 0, 9)
        print(f"Successfully loaded {len(self.labels)} rows!")
        print(f"Label range: min={self.labels.min()}, max={self.labels.max()}")

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        img_flat = np.array(self.pixels[idx], dtype=np.float32)
        num_features = len(img_flat)

        try:
            if num_features >= 3072:
                img = img_flat[:3072].reshape(32, 32, 3)
            else:
                side = max(int(np.sqrt(max(num_features // 3, 1))), 1)
                img = np.resize(img_flat, (side, side, 3))
        except Exception:
            img = np.resize(img_flat, (32, 32, 3))

        img = np.clip(img, 0, 255).astype(np.uint8)

        if self.transform:
            img = self.transform(img)

        label = self.labels[idx]
        return img, label

# Defining standard CIFAR-10 image transforms
transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((32, 32)),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
])

# Instantiate dataset and dataloader
dataset = RobustCSVDataset('train (1).csv', transform=transform)
dataloader = DataLoader(dataset, batch_size=128, shuffle=True, num_workers=2)

# verification test
images, labels = next(iter(dataloader))
print("Successfully loaded batch!")
print("Batch image shape:", images.shape)
print("Batch labels shape:", labels.shape)
```

**3. Baseline Model Development:**

```python
# Splitting dataset into Train (80%) and Validation (20%)
val_size = int(0.2 * len(dataset))
train_size = len(dataset) - val_size
train_ds, val_ds = random_split(dataset, [train_size, val_size])

train_loader = DataLoader(train_ds, batch_size=128, shuffle=True, num_workers=2)
val_loader = DataLoader(val_ds, batch_size=128, shuffle=False, num_workers=2)

print(f"Training samples: {train_size}, Validation samples: {val_size}")

# Defining Baseline CNN Architecture
class BaselineCNN(nn.Module):
    def __init__(self, num_classes=10):
        super(BaselineCNN, self).__init__()
        self.conv1 = nn.Conv2d(3, 32, 3, padding=1)
        self.bn1 = nn.BatchNorm2d(32)
        self.conv2 = nn.Conv2d(32, 64, 3, padding=1)
        self.bn2 = nn.BatchNorm2d(64)
        self.conv3 = nn.Conv2d(64, 128, 3, padding=1)
        self.bn3 = nn.BatchNorm2d(128)
        self.pool = nn.MaxPool2d(2, 2)
        self.fc1 = nn.Linear(128 * 4 * 4, 512)
        self.dropout = nn.Dropout(0.5)
        self.fc2 = nn.Linear(512, num_classes)

    def forward(self, x):
        x = self.pool(F.relu(self.bn1(self.conv1(x))))
        x = self.pool(F.relu(self.bn2(self.conv2(x))))
        x = self.pool(F.relu(self.bn3(self.conv3(x))))
        x = x.view(-1, 128 * 4 * 4)
        x = self.dropout(F.relu(self.fc1(x)))
        x = self.fc2(x)
        return x

model = BaselineCNN(num_classes=10).to(device)

# Loss / Optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

#  Training Loop
epochs = 10
for epoch in range(epochs):
    model.train()
    running_loss, correct, total = 0.0, 0, 0

    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()

    train_loss = running_loss / total
    train_acc = correct / total

    # Validation Phase
    model.eval()
    val_loss, val_correct, val_total = 0.0, 0, 0
    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)

            val_loss += loss.item() * inputs.size(0)
            _, predicted = outputs.max(1)
            val_total += labels.size(0)
            val_correct += predicted.eq(labels).sum().item()

    v_loss = val_loss / val_total
    v_acc = val_correct / val_total
    print(f"Epoch [{epoch+1}/{epochs}] | Train Loss: {train_loss:.4f} | Train Acc: {train_acc*100:.2f}% | Val Loss: {v_loss:.4f} | Val Acc: {v_acc*100:.2f}%")

# Saving Model and Evaluating Baseline Metrics
torch.save(model.state_dict(), "baseline_model.pth")
model_size_mb = os.path.getsize("baseline_model.pth") / (1024 * 1024)
print(f"\nBaseline Model Size: {model_size_mb:.2f} MB")

# Inference Latency Benchmark
model.eval()
dummy_input = torch.randn(1, 3, 32, 32).to(device)
for _ in range(10):
    _ = model(dummy_input)

start_time = time.time()
iterations = 500
with torch.no_grad():
    for _ in range(iterations):
        _ = model(dummy_input)
end_time = time.time()

latency_ms = ((end_time - start_time) * 1000) / iterations
print(f"Average Inference Latency: {latency_ms:.2f} ms per image")
```

**4. Apply Model Compression Techniques:**

```python
# Implementing Weight Pruning
import torch.nn.utils.prune as prune
import copy

# Loading saved baseline model
pruned_model = BaselineCNN(num_classes=10).to(device)
pruned_model.load_state_dict(torch.load("baseline_model.pth"))

# Applying L1 Unstructured Pruning
for name, module in pruned_model.named_modules():
    if isinstance(module, (nn.Conv2d, nn.Linear)):
        prune.l1_unstructured(module, name='weight', amount=0.3)
        # Making pruning permanent so the zeros stay
        prune.remove(module, 'weight')

print("Successfully applied 30% L1 unstructured pruning!")

#  Quick Fine-Tuning to recover accuracy after pruning
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(pruned_model.parameters(), lr=0.0001)

pruned_model.train()
for inputs, labels in train_loader:
    inputs, labels = inputs.to(device), labels.to(device)
    optimizer.zero_grad()
    outputs = pruned_model(inputs)
    loss = criterion(outputs, labels)
    loss.backward()
    optimizer.step()

# Saving Pruned Model and Measure Size
torch.save(pruned_model.state_dict(), "pruned_model.pth")
pruned_size_mb = os.path.getsize("pruned_model.pth") / (1024 * 1024)
print(f"Pruned Model Size: {pruned_size_mb:.2f} MB")

# Evaluating Pruned Model Validation Accuracy
pruned_model.eval()
val_correct, val_total = 0, 0
with torch.no_grad():
    for inputs, labels in val_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        outputs = pruned_model(inputs)
        _, predicted = outputs.max(1)
        val_total += labels.size(0)
        val_correct += predicted.eq(labels).sum().item()

pruned_acc = (val_correct / val_total) * 100
print(f"Pruned Validation Accuracy: {pruned_acc:.2f}%")

# Evaluating Pruned Model Latency
dummy_input = torch.randn(1, 3, 32, 32).to(device)
for _ in range(10):  # Warm-up
    _ = pruned_model(dummy_input)

start_time = time.time()
with torch.no_grad():
    for _ in range(500):
        _ = pruned_model(dummy_input)
end_time = time.time()

pruned_latency = ((end_time - start_time) * 1000) / 500
print(f"Pruned Inference Latency: {pruned_latency:.2f} ms per image")
```

After applying pruning, we notice that the validation accuracy ticked up slightly to 97.11%, and latency dropped down to 0.61 ms per image.

```python
# Post-Training Dynamic Quantization

import torch
import torch.quantization
import os
import time

#  Loading the baseline model specifically for CPU inference

quant_model = BaselineCNN(num_classes=10).to('cpu')
quant_model.load_state_dict(torch.load("baseline_model.pth", map_location='cpu'))
quant_model.eval()

# Applying Dynamic Quantization to Conv2d and Linear layers
quantized_model = torch.quantization.quantize_dynamic(
    quant_model,
    {nn.Conv2d, nn.Linear},
    dtype=torch.qint8
)

print("Successfully applied dynamic quantization (int8)!")

# Savv Quantized Model and Measuring its Size
torch.save(quantized_model.state_dict(), "quantized_model.pth")
quant_size_mb = os.path.getsize("quantized_model.pth") / (1024 * 1024)
print(f"Quantized Model Size: {quant_size_mb:.2f} MB")

# Evaluating Quantized Model Validation Accuracy
val_correct, val_total = 0, 0
with torch.no_grad():
    for inputs, labels in val_loader:
        # Move inputs to CPU for quantized evaluation
        inputs, labels = inputs.to('cpu'), labels.to('cpu')
        outputs = quantized_model(inputs)
        _, predicted = outputs.max(1)
        val_total += labels.size(0)
        val_correct += predicted.eq(labels).sum().item()

quant_acc = (val_correct / val_total) * 100
print(f"Quantized Validation Accuracy: {quant_acc:.2f}%")

# Evaluating Quantized Model Latency (on CPU)
dummy_input_cpu = torch.randn(1, 3, 32, 32).to('cpu')
for _ in range(10):  # Warm-up
    _ = quantized_model(dummy_input_cpu)

start_time = time.time()
with torch.no_grad():
    for _ in range(500):
        _ = quantized_model(dummy_input_cpu)
end_time = time.time()

quant_latency = ((end_time - start_time) * 1000) / 500
print(f"Quantized Inference Latency: {quant_latency:.2f} ms per image (measured on CPU)")
```

Our quantized model dropped the storage size down to 1.38 MB while keeping its validation accuracy at 96.92%.

**5. Knowledge Distillation:**

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import os
import time

# Defining a smaller Student CNN Architecture
class StudentCNN(nn.Module):
    def __init__(self, num_classes=10):
        super(StudentCNN, self).__init__()
        # Fewer channels and layers than the teacher
        self.conv1 = nn.Conv2d(3, 16, 3, padding=1)
        self.bn1 = nn.BatchNorm2d(16)
        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.bn2 = nn.BatchNorm2d(32)
        self.pool = nn.MaxPool2d(2, 2)
        self.fc1 = nn.Linear(32 * 8 * 8, 128)
        self.fc2 = nn.Linear(128, num_classes)

    def forward(self, x):
        x = self.pool(F.relu(self.bn1(self.conv1(x))))
        x = self.pool(F.relu(self.bn2(self.conv2(x))))
        x = x.view(-1, 32 * 8 * 8)
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x

student_model = StudentCNN(num_classes=10).to(device)

# Loading teacher model (Baseline)
teacher_model = BaselineCNN(num_classes=10).to(device)
teacher_model.load_state_dict(torch.load("baseline_model.pth"))
teacher_model.eval()

# Defining Distillation Loss Function
def distillation_loss(student_logits, labels, teacher_logits, alpha=0.5, temperature=4.0):
    ce_loss = F.cross_entropy(student_logits, labels)

    # Softening probabilities with temperature
    soft_student = F.log_softmax(student_logits / temperature, dim=1)
    soft_teacher = F.softmax(teacher_logits / temperature, dim=1)

    kl_div = nn.KLDivLoss(reduction='batchmean')(soft_student, soft_teacher) * (temperature ** 2)

    return alpha * ce_loss + (1 - alpha) * kl_div

optimizer = optim.Adam(student_model.parameters(), lr=0.001)

# Training Student via Distillation (5 Epochs)
print("Training Student Model via Knowledge Distillation...")
for epoch in range(5):
    student_model.train()
    running_loss, correct, total = 0.0, 0, 0

    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        student_logits = student_model(inputs)

        with torch.no_grad():
            teacher_logits = teacher_model(inputs)

        loss = distillation_loss(student_logits, labels, teacher_logits, alpha=0.3, temperature=4.0)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)
        _, predicted = student_logits.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()

    print(f"Epoch [{epoch+1}/5] | Distillation Loss: {running_loss/total:.4f} | Student Acc: {(correct/total)*100:.2f}%")

# Saving and Evaluating Student Metrics
torch.save(student_model.state_dict(), "student_model.pth")
student_size_mb = os.path.getsize("student_model.pth") / (1024 * 1024)
print(f"\nStudent Model Size: {student_size_mb:.2f} MB")

# Latency Benchmark
student_model.eval()
dummy_input = torch.randn(1, 3, 32, 32).to(device)
for _ in range(10): _ = student_model(dummy_input)

start_time = time.time()
with torch.no_grad():
    for _ in range(500): _ = student_model(dummy_input)
end_time = time.time()

student_latency = ((end_time - start_time) * 1000) / 500
print(f"Student Inference Latency: {student_latency:.2f} ms per image")
```

**6. Low-Rank Factorization:**

```python
import numpy as np

# Loading baseline model for factorization
factorized_model = BaselineCNN(num_classes=10).to(device)
factorized_model.load_state_dict(torch.load("baseline_model.pth"))

# Applying SVD-based low-rank approximation to Conv2d layers
print("Applying low-rank factorization to convolutional layers...")
with torch.no_grad():
    for name, module in factorized_model.named_modules():
        if isinstance(module, nn.Conv2d):
            weight = module.weight.data # Shape: [C_out, C_k, k_h, k_w]
            C_out, C_in, k_h, k_w = weight.shape

            # Reshape to 2D matrix for SVD
            weight_2d = weight.view(C_out, -1)

            #  SVD
            U, S, V = torch.linalg.svd(weight_2d, full_matrices=False)

            # Keepv top rank components
            rank = max(int(len(S) * 0.75), 1)
            U_r = U[:, :rank]
            S_r = S[:rank]
            V_r = V[:rank, :]

            # Reconstruct low-rank weight matrix
            approx_weight_2d = torch.matmul(U_r * S_r, V_r)
            module.weight.data = approx_weight_2d.view(C_out, C_in, k_h, k_w)

print("Low-rank factorization applied successfully!")

#  Fine-tune factorized model (1 epoch) to recover accuracy
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(factorized_model.parameters(), lr=0.0001)

factorized_model.train()
for inputs, labels in train_loader:
    inputs, labels = inputs.to(device), labels.to(device)
    optimizer.zero_grad()
    outputs = factorized_model(inputs)
    loss = criterion(outputs, labels)
    loss.backward()
    optimizer.step()

# Saving and Evaluating Factorized Model Metrics
torch.save(factorized_model.state_dict(), "factorized_model.pth")
factorized_size_mb = os.path.getsize("factorized_model.pth") / (1024 * 1024)
print(f"\nFactorized Model Size: {factorized_size_mb:.2f} MB")

# Evaluating Accuracy
factorized_model.eval()
val_correct, val_total = 0, 0
with torch.no_grad():
    for inputs, labels in val_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        outputs = factorized_model(inputs)
        _, predicted = outputs.max(1)
        val_total += labels.size(0)
        val_correct += predicted.eq(labels).sum().item()

factorized_acc = (val_correct / val_total) * 100
print(f"Factorized Validation Accuracy: {factorized_acc:.2f}%")

# Evaluating Latency
dummy_input = torch.randn(1, 3, 32, 32).to(device)
for _ in range(10): _ = factorized_model(dummy_input)

start_time = time.time()
with torch.no_grad():
    for _ in range(500): _ = factorized_model(dummy_input)
end_time = time.time()

factorized_latency = ((end_time - start_time) * 1000) / 500
print(f"Factorized Inference Latency: {factorized_latency:.2f} ms per image")
```
