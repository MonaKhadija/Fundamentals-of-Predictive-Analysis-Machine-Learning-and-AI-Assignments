**1. Data Preparation:**

We will use PyTorch and the CIFAR-10 dataset.

Environment setup and Dataset Splitting:

```python
# Importing libraries
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import Subset, DataLoader
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

```python
# Preprocessing transforms
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010))
])

# Downloading full CIFAR-10 dataset
trainset_full = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform)
testset_full = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)
```

```python
# Helper function to filter datasets by specific classes
def filter_dataset_by_classes(dataset, classes):
    indices = [i for i, (_, label) in enumerate(dataset) if label in classes]
    return Subset(dataset, indices)

# Defining Subsets
# Subset A: Classes 0 to 4 (e.g., airplane, automobile, bird, cat, deer)
# Subset B: Classes 5 to 9 (e.g., dog, frog, horse, ship, truck)
classes_a = [0, 1, 2, 3, 4]
classes_b = [5, 6, 7, 8, 9]

train_subset_a = filter_dataset_by_classes(trainset_full, classes_a)
train_subset_b = filter_dataset_by_classes(trainset_full, classes_b)

test_subset_a = filter_dataset_by_classes(testset_full, classes_a)
test_subset_b = filter_dataset_by_classes(testset_full, classes_b)

train_loader_a = DataLoader(train_subset_a, batch_size=64, shuffle=True)
train_loader_b = DataLoader(train_subset_b, batch_size=64, shuffle=True)
test_loader_a = DataLoader(test_subset_a, batch_size=64, shuffle=False)
test_loader_b = DataLoader(test_subset_b, batch_size=64, shuffle=False)

print(f"Subset A training samples: {len(train_subset_a)}")
print(f"Subset B training samples: {len(train_subset_b)}")
```

**2. Initial Model Training:**

```python
# Defining CNN and Train on Subset A
class SimpleCNN(nn.Module):
    def __init__(self, num_classes=5):
        super(SimpleCNN, self).__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2)
        )
        self.classifier = nn.Sequential(
            nn.Linear(64 * 8 * 8, 128),
            nn.ReLU(),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = x.view(x.size(0), -1)
        x = self.classifier(x)
        return x

def train_model(model, train_loader, criterion, optimizer, epochs=3):
    model.train()
    for epoch in range(epochs):
        running_loss = 0.0
        for images, labels in train_loader:

            labels = labels % 5
            images, labels = images.to(device), labels.to(device)

            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item()
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {running_loss/len(train_loader):.4f}")

def evaluate_model(model, test_loader):
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in test_loader:
            target_labels = labels % 5
            images, target_labels = images.to(device), target_labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs.data, 1)
            total += target_labels.size(0)
            correct += (predicted == target_labels).sum().item()
    accuracy = 100 * correct / total
    return accuracy

# Initializing and training initial model on Subset A
initial_model = SimpleCNN(num_classes=5).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(initial_model.parameters(), lr=0.001)

print("--- Training Initial Model on Subset A ---")
train_model(initial_model, train_loader_a, criterion, optimizer, epochs=5)

acc_a = evaluate_model(initial_model, test_loader_a)
print(f"Initial Model Accuracy on Subset A Test Set: {acc_a:.2f}%")
```

**3. Simulating Data Distribution Shift:**

```python
# Evaluating initial model on Subset B without retraining
acc_b_initial = evaluate_model(initial_model, test_loader_b)
print(f"Initial Model Accuracy on Subset B Test Set: {acc_b_initial:.2f}%")
```

Note: we can clearly notice that accuracy on Subset B is near random guessing because the model has never encountered dogs, frogs, horses, ships, or trucks.

**4. Implementing Continual Learning Techniques:**


```python
# Combining datasets for Stateless Retraining
train_subset_ab = torch.utils.data.ConcatDataset([train_subset_a, train_subset_b])
test_subset_ab = torch.utils.data.ConcatDataset([test_subset_a, test_subset_b])

train_loader_ab = DataLoader(train_subset_ab, batch_size=64, shuffle=True)
test_loader_ab = DataLoader(test_subset_ab, batch_size=64, shuffle=False)

# Stateless Retraining (From scratch on A + B)

stateless_model = SimpleCNN(num_classes=10).to(device)
criterion_full = nn.CrossEntropyLoss()
optimizer_stateless = optim.Adam(stateless_model.parameters(), lr=0.001)

print("--- Stateless Retraining (A + B from scratch) ---")
# Custom loop because full dataset uses labels 0-9 directly
stateless_model.train()
for epoch in range(5):
    for images, labels in train_loader_ab:
        images, labels = images.to(device), labels.to(device)
        optimizer_stateless.zero_grad()
        outputs = stateless_model(images)
        loss = criterion_full(outputs, labels)
        loss.backward()
        optimizer_stateless.step()

# Stateful Training

stateful_model = initial_model # continuing from initial_model
# Replace final layer to accommodate 10 classes while keeping feature extractor weights
stateful_model.classifier[2] = nn.Linear(128, 10).to(device)
optimizer_stateful = optim.Adam(stateful_model.parameters(), lr=0.0005)

print("--- Stateful Training (Fine-tuning on Subset B) ---")
stateful_model.train()
for epoch in range(5):
    for images, labels in train_loader_b:
        images, labels = images.to(device), labels.to(device)
        optimizer_stateful.zero_grad()
        outputs = stateful_model(images)
        loss = criterion_full(outputs, labels)
        loss.backward()
        optimizer_stateful.step()

# Evaluation helper for 10-class models
def evaluate_full_model(model, test_loader):
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in test_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    return 100 * correct / total

print(f"Stateless Model Accuracy (Combined A+B): {evaluate_full_model(stateless_model, test_loader_ab):.2f}%")
print(f"Stateful Model Accuracy (Combined A+B): {evaluate_full_model(stateful_model, test_loader_ab):.2f}%")
```

**5. Implementing Evaluation Techniques:**

```python
# A/B Testing Simulation
print("--- A/B Testing Simulation ---")
group_a_correct, group_b_correct = 0, 0
total_requests = 500

# Randomly sample requests from test_loader_ab
test_samples = list(test_subset_ab)
sampled_requests = random.sample(test_samples, total_requests)

for image, label in sampled_requests[:total_requests//2]:
    img_tensor = image.unsqueeze(0).to(device)
    output = stateless_model(img_tensor)
    pred = output.argmax(dim=1).item()
    if pred == label: group_a_correct += 1

for image, label in sampled_requests[total_requests//2:]:
    img_tensor = image.unsqueeze(0).to(device)
    output = stateful_model(img_tensor)
    pred = output.argmax(dim=1).item()
    if pred == label: group_b_correct += 1

print(f"Group A (Stateless Model) Accuracy: {(group_a_correct / (total_requests//2)) * 100:.2f}%")
print(f"Group B (Stateful Model) Accuracy: {(group_b_correct / (total_requests//2)) * 100:.2f}%")

# Epsilon-Greedy Bandit Algorithm for Model Selection
print("\n--- Multi-Armed Bandit Model Selection ($\epsilon$-Greedy) ---")
epsilon = 0.1  # Exploration rate
arms = {'stateless': {'pulls': 0, 'rewards': 0.0}, 'stateful': {'pulls': 0, 'rewards': 0.0}}

bandit_correct = 0
for image, label in sampled_requests:
    img_tensor = image.unsqueeze(0).to(device)

    # Epsilon-greedy choice
    if random.random() < epsilon:
        chosen_model_name = random.choice(['stateless', 'stateful'])
    else:
        # Exploit: choose model with higher average reward
        avg_rewards = {name: (data['rewards'] / data['pulls'] if data['pulls'] > 0 else 0) for name, data in arms.items()}
        chosen_model_name = max(avg_rewards, key=avg_rewards.get)

    # Inference
    model_to_use = stateless_model if chosen_model_name == 'stateless' else stateful_model
    pred = model_to_use(img_tensor).argmax(dim=1).item()

    reward = 1.0 if pred == label else 0.0
    bandit_correct += reward

    # Update arm statistics
    arms[chosen_model_name]['pulls'] += 1
    arms[chosen_model_name]['rewards'] += reward

print(f"Bandit Strategy Overall Accuracy: {(bandit_correct / total_requests) * 100:.2f}%")
print(f"Arm Pull Statistics: {arms}")
```

**Reporting and Analysis:**

This project simulates a production-like environment using the CIFAR-10 dataset divided into two distinct subsets—Subset A and Subset B. Tese sbsets are devided as followed: Subset A represent the initial operational environment (classes 0 to 4: airplane, automobile, bird, cat, deer). Subset B represent the introduction of a new data distribution (classes 5 to 9: dog, frog, horse, ship, truck).

When the model was trained exclusively on Subset A, it caused a silent Degradation when evaluated on subset B, highlighting the need for automated drift detection and continual updating mechanisms.

To adapt to the new classes, two distinct approaches were implemented and compared: Stateless Retraining that achieved strong overall accuracy, then Stateful Training that achieved lower overall accuracy.

To safely manage model updates in a live environment, traffic simulation techniques were deployed: A/B Testing Simulation resultig in Group A achieving 74.40% accuracy, while Group B stayed behind at 44.00%, confirming that stateless retraining shows superior reliability for this specific setup.
The bandit strategy achieved an overall accuracy of 73.00%.The bandit algorithm successfully minimized exposure to the inferior stateful model, automatically optimizing traffic flow and protecting user experience without manual intervention.

Takeaways: Stateless retraining is more resilient against catastrophic forgetting than fine-tuning, while multi-armed bandits offer a safety net for production routing.
