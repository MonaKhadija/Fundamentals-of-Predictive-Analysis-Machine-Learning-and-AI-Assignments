**1. Load and Explore a Dataset:**

```python
# we will be loading the Wine dataset from Scikit-Learn
# We start by importing the needed libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.datasets import load_wine
```

```python
# Load data
wine = load_wine()
df = pd.DataFrame(wine.data, columns=wine.feature_names)
df['target'] = wine.target

print(f"Dataset Shape: {df.shape}")
print("Class Distribution:\n", df['target'].value_counts())
```

This dataset has 178 entries, 13 chemical attributes, and 3 distinct classes.

**2. Apply Sampling Techniques:**

In this step, we will perform random vs. stratified sampling and handle class imbalance using SMOTE.

```python
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
```

```python
# Sampling with 50 samples
random_sample = df.sample(n=50, random_state=42)
stratified_sample = df.groupby('target', group_keys=False).apply(
    lambda x: x.sample(n=16, random_state=42)
)

# Imbalance & SMOTE
imbalanced_df = df[~(df['target'] == 2)].copy()  # Remove class 2 partially
X, y = imbalanced_df.drop('target', axis=1), imbalanced_df['target']
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)
```

```python
import matplotlib.pyplot as plt

# Visualizing class distributions side-by-side
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

random_sample['target'].value_counts().sort_index().plot(
    kind='bar', ax=axes[0], color='skyblue'
)
axes[0].set_title('Random Sampling Distribution')
axes[0].set_xlabel('Class')
axes[0].set_ylabel('Frequency')

stratified_sample['target'].value_counts().sort_index().plot(
    kind='bar', ax=axes[1], color='salmon'
)
axes[1].set_title('Stratified Sampling Distribution')
axes[1].set_xlabel('Class')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

We can notice that Random Sampling shows an uneven representation of classes, while Stratified Sampling Shows a proportionally balanced count for each class.

**3. Labeling Data Using Heuristics:**

We will be creating a rule-based label and evaluating a Decision Tree using ROC-AUC.

```python
from sklearn.metrics import auc, roc_curve
from sklearn.tree import DecisionTreeClassifier

# Heuristic labeling based on alcohol threshold
df['heuristic_label'] = np.where(df['alcohol'] > 13.0, 1, 0)

# Decision Tree & ROC-AUC
X_orig, y_orig = df[wine.feature_names], df['target']
X_train, X_test, y_train, y_test = train_test_split(
    X_orig, (y_orig == 0).astype(int), test_size=0.3, random_state=42
)
dt = DecisionTreeClassifier(random_state=42).fit(X_train, y_train)
y_prob = dt.predict_proba(X_test)[:, 1]
fpr, tpr, _ = roc_curve(y_test, y_prob)
print(f"AUC Score: {auc(fpr, tpr):.2f}")
```

Heuristic probable issues: Rule-based labels introduce noise because real-world thresholds don't capture complex patterns easily.

ROC Curve: Plots True Positive Rate against False Positive Rate.

**4. Data Augmentation:**

we will Add Gaussian noise to simulate data augmentation and compare Logistic Regression performance.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Data Augmentation (adding small random noise)
noise = np.random.normal(0, 0.1, X_train.shape)
X_train_augmented = X_train + noise

# Train & Evaluate
lr_orig = LogisticRegression(max_iter=1000).fit(X_train, y_train)
lr_aug = LogisticRegression(max_iter=1000).fit(X_train_augmented, y_train)

print(
    'Original Accuracy:',
    accuracy_score(y_test, lr_orig.predict(X_test)),
)
print('Augmented Accuracy:', accuracy_score(y_test, lr_aug.predict(X_test)))
```

Augmentation usually has an Impact: Controlled noise acts as regularization, helping prevent overfitting on small datasets.
